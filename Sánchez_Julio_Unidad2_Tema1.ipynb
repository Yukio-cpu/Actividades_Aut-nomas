### Ejercicio 1
**Programa que permita registrar las notas de varios estudiantes y calcular el promedio de las notas ingresadas.**

Requisitos:
- Bucle do-while (simulado con `while True`).
- Funciones sobrecargadas para calcular el promedio con notas `int` y `float`.
- El usuario elige el tipo de nota a ingresar.
- Al finalizar: mostrar cantidad total de notas y el promedio.

````python
# ============================================================
# EJERCICIO 1 — Registro de notas y cálculo de promedio
# Sobrecarga de funciones simulada mediante despacho por tipo
# ============================================================

def calcular_promedio_int(notas: list) -> int:
    """Calcula el promedio de una lista de notas enteras (int)."""
    if len(notas) == 0:
        return 0
    return int(sum(notas) // len(notas))   # División entera para mantener tipo int


def calcular_promedio_float(notas: list) -> float:
    """Calcula el promedio de una lista de notas decimales (float)."""
    if len(notas) == 0:
        return 0.0
    return sum(notas) / len(notas)          # División real para mantener precisión float


def registrar_notas():
    """Función principal: controla el ingreso de notas con bucle do-while."""
    notas_int   = []   # Lista para notas de tipo entero
    notas_float = []   # Lista para notas de tipo decimal

    print("=== Registro de Notas de Estudiantes ===")

    # --- Bucle do-while (Python no tiene do-while nativo; se simula con while True) ---
    while True:
        print("\nOpciones:")
        print("  1 — Ingresar nota entera (int)")
        print("  2 — Ingresar nota decimal (float)")
        print("  0 — Finalizar ingreso")

        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            try:
                nota = int(input("Ingrese la nota entera: "))  # Conversión explícita a int
                notas_int.append(nota)
                print(f"  Nota {nota} registrada (int).")
            except ValueError:
                print("  Error: ingrese un número entero válido.")

        elif opcion == "2":
            try:
                nota = float(input("Ingrese la nota decimal: "))  # Conversión explícita a float
                notas_float.append(nota)
                print(f"  Nota {nota:.2f} registrada (float).")
            except ValueError:
                print("  Error: ingrese un número decimal válido.")

        elif opcion == "0":
            break   # Condición de salida del bucle
        else:
            print("  Opción inválida. Intente de nuevo.")

    # --- Cálculo y presentación de resultados al finalizar ---
    total = len(notas_int) + len(notas_float)   # Total de notas ingresadas
    print(f"\n{'='*40}")
    print(f"Total de notas ingresadas: {total}")

    if notas_int:
        prom_int = calcular_promedio_int(notas_int)  # Llamada a función sobrecargada int
        print(f"Promedio de notas enteras  : {prom_int} (int)")

    if notas_float:
        prom_float = calcular_promedio_float(notas_float)  # Llamada a función sobrecargada float
        print(f"Promedio de notas decimales: {prom_float:.2f} (float)")

    if not notas_int and not notas_float:
        print("No se ingresaron notas.")


# Punto de entrada
registrar_notas()
````

### Ejercicio 2
**Conversor de cantidad de dinero a monedas de diferentes denominaciones.**

Denominaciones: 1 dólar, 50, 25, 10, 5 y 1 centavo.

````python
# ============================================================
# EJERCICIO 2 — Conversor de dinero a monedas
# ============================================================

def calcular_monedas(cantidad_dolares: float) -> dict:
    """
    Recibe una cantidad en dólares y devuelve el número de monedas
    de cada denominación usando el algoritmo greedy (voraz).
    """
    # Convertir a centavos para evitar errores de punto flotante
    centavos = round(cantidad_dolares * 100)

    # Denominaciones en centavos (de mayor a menor)
    denominaciones = [
        (100, "1 dólar   (100¢)"),
        ( 50, "50 centavos"),
        ( 25, "25 centavos"),
        ( 10, "10 centavos"),
        (  5, " 5 centavos"),
        (  1, " 1 centavo "),
    ]

    resultado = {}   # Diccionario: nombre -> cantidad de monedas

    for valor, nombre in denominaciones:
        cantidad = centavos // valor    # Cuántas monedas caben de esta denominación
        centavos %= valor              # Residuo para la siguiente denominación
        resultado[nombre] = cantidad

    return resultado


def mostrar_monedas():
    """Función de interfaz: solicita entrada y muestra los resultados."""
    print("=== Conversor de Dinero a Monedas ===")

    try:
        cantidad = float(input("Ingrese la cantidad en dólares (ej. 2.87): $"))
        if cantidad < 0:
            print("Error: la cantidad no puede ser negativa.")
            return
    except ValueError:
        print("Error: ingrese un valor numérico válido.")
        return

    monedas = calcular_monedas(cantidad)   # Llamada a la función principal

    print(f"\n${cantidad:.2f} se puede dividir en:")
    print("-" * 35)
    for denominacion, cantidad_monedas in monedas.items():
        if cantidad_monedas > 0:   # Solo mostrar denominaciones utilizadas
            print(f"  {denominacion}: {cantidad_monedas} moneda(s)")


mostrar_monedas()
````

### Ejercicio 3
**Programa que resuelve ecuaciones cuadráticas de la forma ax² + bx + c = 0.**

Requisitos: función para el discriminante, función para determinar el tipo de solución, mostrar el discriminante antes de las raíces.

````python
# ============================================================
# EJERCICIO 3 — Resolución de ecuaciones cuadráticas
# ============================================================

import math   # Para calcular la raíz cuadrada


def calcular_discriminante(a: float, b: float, c: float) -> float:
    """Calcula el discriminante D = b² - 4ac."""
    return b**2 - 4 * a * c    # Fórmula del discriminante


def resolver_ecuacion(a: float, b: float, c: float, D: float):
    """
    Determina el tipo de soluciones según el discriminante D
    y calcula las raíces si son reales.
    """
    if D > 0:
        # Dos soluciones reales y distintas
        x1 = (-b + math.sqrt(D)) / (2 * a)   # Fórmula cuadrática (+)
        x2 = (-b - math.sqrt(D)) / (2 * a)   # Fórmula cuadrática (-)
        print("Tipo de solución: Dos raíces reales y distintas (D > 0)")
        print(f"  x1 = {x1:.2f}")
        print(f"  x2 = {x2:.2f}")

    elif D == 0:
        # Una solución real doble
        x = -b / (2 * a)    # Raíz única cuando el discriminante es cero
        print("Tipo de solución: Una raíz real doble (D = 0)")
        print(f"  x = {x:.2f}")

    else:
        # Dos soluciones complejas (D < 0)
        print("Tipo de solución: Dos raíces complejas (D < 0)")
        print("  La ecuación no tiene soluciones reales.")


def ecuacion_cuadratica():
    """Función principal: solicita coeficientes y coordina el cálculo."""
    print("=== Resolución de Ecuaciones Cuadráticas (ax² + bx + c = 0) ===")

    try:
        a = float(input("Ingrese el coeficiente a: "))
        b = float(input("Ingrese el coeficiente b: "))
        c = float(input("Ingrese el coeficiente c: "))
    except ValueError:
        print("Error: ingrese valores numéricos válidos.")
        return

    if a == 0:
        print("Error: el coeficiente 'a' no puede ser cero (no sería cuadrática).")
        return

    D = calcular_discriminante(a, b, c)    # Calcular discriminante

    print(f"\nEcuación: {a}x² + {b}x + {c} = 0")
    print(f"Discriminante calculado: D = {D:.2f}")   # Mostrar D antes de las soluciones
    print()

    resolver_ecuacion(a, b, c, D)   # Determinar y mostrar soluciones


ecuacion_cuadratica()
````

### Ejercicio 4
**Conversor de hora en formato 24 horas a formato 12 horas con indicación AM/PM.**

Requisitos: validación de rango, funciones, condicionales anidados, manejo de medianoche y mediodía, sin arreglos ni cadenas de texto para la conversión.

````python
# ============================================================
# EJERCICIO 4 — Conversor de formato 24h a 12h con AM/PM
# No se usan listas ni cadenas para la lógica de conversión.
# ============================================================

def validar_hora(hora: int, minutos: int) -> bool:
    """Valida que la hora esté en [0-23] y los minutos en [0-59]."""
    if hora < 0 or hora > 23:       # Verificar rango de horas
        return False
    if minutos < 0 or minutos > 59: # Verificar rango de minutos
        return False
    return True


def determinar_periodo(hora: int) -> int:
    """
    Determina el periodo: retorna 0 para AM, 1 para PM.
    Resuelto únicamente con condicionales, sin cadenas de texto.
    """
    if hora < 12:     # Horas 0-11 corresponden a AM
        return 0      # 0 representa AM
    else:             # Horas 12-23 corresponden a PM
        return 1      # 1 representa PM


def convertir_hora(hora: int) -> int:
    """
    Convierte una hora en formato 24h al equivalente en 12h.
    Maneja los casos especiales de medianoche (0) y mediodía (12).
    """
    if hora == 0:       # Medianoche: 0:00 → 12:XX AM
        return 12
    elif hora <= 12:    # De 1:00 a 12:00 no se modifica
        return hora
    else:               # De 13:00 a 23:00 se resta 12
        return hora - 12


def imprimir_periodo(periodo: int):
    """Imprime AM o PM según el valor del periodo (0=AM, 1=PM)."""
    if periodo == 0:
        print("AM", end="")   # Condicional puro sin cadenas en la lógica
    else:
        print("PM", end="")


def convertir_formato():
    """Función principal: controla la entrada, validación y presentación."""
    print("=== Conversor de Hora 24h a 12h ===")

    try:
        hora    = int(input("Ingrese la hora    (0-23): "))
        minutos = int(input("Ingrese los minutos (0-59): "))
    except ValueError:
        print("Error: ingrese valores enteros válidos.")
        return

    # Validar rango antes de continuar
    if not validar_hora(hora, minutos):
        print("Error: hora o minutos fuera del rango permitido. Programa finalizado.")
        return

    hora_12  = convertir_hora(hora)       # Conversión de la hora
    periodo  = determinar_periodo(hora)   # Determinar AM o PM

    # Mostrar hora original en formato 24h
    print(f"\nHora ingresada (24h): {hora:02d}:{minutos:02d}")

    # Mostrar hora convertida en formato 12h
    print(f"Hora convertida (12h): {hora_12:02d}:{minutos:02d} ", end="")
    imprimir_periodo(periodo)   # Imprimir AM/PM con función pura de condicionales
    print()   # Salto de línea final


convertir_formato()
````

### 1. Comprensión
**¿Por qué el uso de funciones mejora la legibilidad y el mantenimiento de un programa?**

Cuando un programa crece, el código escrito de forma lineal se vuelve difícil de leer y de corregir. Las funciones permiten dividir ese código en bloques con nombres descriptivos, cada uno responsable de una tarea específica. Al leer el programa principal, es posible entender la lógica general sin conocer todos los detalles internos de cada bloque. Además, si hay que corregir un error o modificar un comportamiento, solo se interviene en la función correspondiente, sin riesgo de afectar el resto del programa.

**Diferencias entre funciones con y sin parámetros:**

Una función sin parámetros siempre realiza exactamente la misma operación cada vez que se llama, porque no recibe datos externos. Por ejemplo, una función `mostrar_menu()` siempre mostrará el mismo menú. Una función con parámetros, en cambio, adapta su comportamiento según los datos que recibe; por ejemplo, `calcular_promedio(notas)` produce resultados distintos según qué lista se le pase.

**Diferencias entre funciones que retornan valor y las que no:**

Una función que retorna un valor (como `calcular_discriminante`) entrega un resultado que puede ser usado en otras operaciones o almacenado en una variable. Una función sin retorno (tipo `void` en C++, o que solo usa `print` en Python) realiza una acción directamente, como mostrar información en pantalla, sin producir un valor reutilizable. En general, las funciones con retorno son más versátiles porque permiten encadenar operaciones.

### 2. Reflexión Personal
**¿Qué ventajas tiene la sobrecarga de funciones frente a crear funciones con diferentes nombres?**

La sobrecarga de funciones permite usar el mismo nombre para operaciones conceptualmente idénticas que difieren solo en el tipo de datos que reciben. Esto hace que el código sea más limpio e intuitivo, ya que el programador no necesita recordar varios nombres como `calcular_promedio_enteros` y `calcular_promedio_decimales`; simplemente llama a `calcular_promedio` y el compilador o el sistema de tipos determina cuál versión aplicar.

**Ejemplo útil:** En el Ejercicio 1, se definieron dos versiones del promedio: una para `int` y otra para `float`. Si existieran tres o cuatro tipos adicionales (como `double` o `long`), mantener nombres distintos generaría confusión. Con sobrecarga, el nombre único `calcular_promedio` actúa como una interfaz coherente independientemente del tipo de entrada.
